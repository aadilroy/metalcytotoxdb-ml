```python
# Paths (relative to the repository root, so the notebooks run on any machine)
# Expected layout:  <repo>/notebooks/ (this file), <repo>/data/, <repo>/models/,
#                   <repo>/figures/, <repo>/results/, <repo>/feature_importance/
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = ROOT / "data"
MODELS_DIR = ROOT / "models"
FIG_DIR = ROOT / "figures"
RESULTS_DIR = ROOT / "results"
FI_DIR = ROOT / "feature_importance"
for _d in (DATA_DIR, MODELS_DIR, FIG_DIR, RESULTS_DIR, FI_DIR):
    _d.mkdir(parents=True, exist_ok=True)
```

```python
import pandas as pd
import numpy as np

from rdkit import Chem

from mordred import Calculator, descriptors
```

```python
df = pd.read_csv(str(DATA_DIR / "MetalCytoToxDB.csv"))
df["row_id"] = df.index

df["Molecule"] = df["SMILES_Ligands"].apply(Chem.MolFromSmiles)
```

```python
#Step 3
#Create the Mordred calculator.

calc = Calculator(descriptors, ignore_3D=True)
```

```python
mordred_df = calc.pandas(df["Molecule"])
```

```python
print(mordred_df.shape)
```

```python
mordred_df.isnull().sum().sort_values(ascending=False).head(20)
```

```python
rdkit_all = pd.read_csv(str(DATA_DIR / "RDKit_Descriptors.csv"))
```

```python
df_all = pd.concat(
    [
        df.reset_index(drop=True),
        rdkit_all.reset_index(drop=True),
        mordred_df.reset_index(drop=True)
    ],
    axis=1
)

print(df_all.shape)
```

```python
df_all.to_csv(str(DATA_DIR / "Descriptors_Full.csv"), index=False)
```

```python
mordred_df.to_csv(str(DATA_DIR / "Mordred_Descriptors.csv"), index=False)
```
